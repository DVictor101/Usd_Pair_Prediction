```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import joblib
```

```python
## loading explored data

data = pd.read_csv("./Data/explored_data.csv")
```

```python
data.head(1)
```

```python
data.columns
```

```python
## viewing missing values

data.isna().sum()
```

```python
## handling missing values(numbers)

numeric_columns = data.select_dtypes(include = np.number).columns

data[numeric_columns] = data[numeric_columns].fillna(data[numeric_columns].median())
```

```python
## handling missing values(categorical)

categorical_columns = data.select_dtypes(include="object").columns

data[categorical_columns] = data[categorical_columns].fillna("Unknown")
```

```python
data.isna().sum()
```

```python
## target variable

target = "Close_48h_after"
```

```python
data.columns
```

```python
# predictor variables

drop_columns = ['Currency_Code', 'DateTime',"Unnamed: 0", 'Volatility_5min', 'Close_24h_after', 'Price_Change_5min', 'Price_Change_24h', 'Close_5min_after', 'Absolute_Move_48h', 'Absolute_Move_24h', 'Absolute_Move_5min', 'Direction_5min', 'Month_Number',"Price_Change_48h", 'Close_48h_after', 'price_change_5min', 'Price_Change_48h', 'Direction_24h', 'Direction_48h', 'news_change', 'news_change_Percentage', 'Move_before_release', 'Move_after_release', 'Hour']
```

```python
## stating the x and y variables

x = data.drop(columns = drop_columns)

y = data[target]
```

```python
##we are training with only the data that will be available at the point of prediction

x.columns
```

```python
## encoding categorical variables



numeric_features = x.select_dtypes(include = np.number).columns

categorical_features = x.select_dtypes(include = "object").columns
```

```python
preprocessor = ColumnTransformer(
    transformers =[
        ("num",
        StandardScaler(),
        numeric_features),
        ("cat",
        OneHotEncoder(handle_unknown="ignore"),
        categorical_features
        )
    ]
)
```

```python
## splitting the data set

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 31)
```

```python
## preprocessing the data

x_train_processed = preprocessor.fit_transform(x_train)
x_test_processed = preprocessor.transform(x_test)
```

```python
# saving the preprocessor

joblib.dump(preprocessor, "./Models/preprocessor.pkl")
```

```python
## saving the processed data

joblib.dump(x_train_processed, "./Models/x_train.pkl")

joblib.dump(x_test_processed, "./Models/x_test.pkl")

joblib.dump(y_train, "./Models/y_train.pkl")

joblib.dump(y_test, "./Models/y_test.pkl")
```
