# 📊 Predicting the Impact of USD and JPY Economic News on the USD/JPY Currency Pair

## 1. Project Overview

The foreign exchange (Forex) market is highly sensitive to macroeconomic news releases. High-impact economic announcements from the United States and Japan, such as interest rate decisions, inflation reports, GDP releases, employment statistics, and central bank communications, often lead to significant movements in the USD/JPY Pair.

This project investigates how historical USD and JPY economic news events influence the price movement of the USD/JPY currency pair. By combining historical economic news data with minute-level market price data, the project measures market reactions before, during, and after each news release.

The insights gained from this analysis will be used to build predictive models capable of estimating possible market reactions to future economic announcements.
    

# 2. Problem Statement

Economic news plays a crucial role in the Forex market, yet traders often rely on experience or manual analysis to estimate its impact. There is limited quantitative analysis that measures how specific USD and JPY news events influence the USD/JPY exchange rate over different time intervals.

This project seeks to bridge that gap by collecting historical economic news and market price data, aligning them based on their release times, and analyzing the resulting price movements to uncover meaningful patterns.


# 3. Project Objectives

* Collect historical USD and JPY economic news.
* Collect historical USD/JPY minute-level price data.
* Clean and preprocess both datasets.
* Align economic news with corresponding market prices.
* Extract the market price:

  * 5 minutes before the news release
  * At the exact release time
  * 5 minutes after the release
  * 24 hours after the release
  * 48 hours after the release
* Analyze how different economic events affect the USD/JPY.
* Build predictive models capable of estimating potential market reactions to future economic news releases.

````python
##importing used packges
import pandas as pd
import numpy as np
````

````python
#importing the news data


data = pd.read_csv("./Data/forex_factory_2024_events.csv")
````

````python
#viewing top 5 of our news data

data.head()
````

````python
## dropping all columns with time == all day

data = data.drop(data[data["time"] == "All Day"].index)
````

````python
#viewing top 5 after dropping columns all day

data.head()
````

````python
# dropping the impact columns because it is not relevant to our project, we want the model to determine the impact 
## dropping month and day

data = data.drop(columns = ["impact", "month", "day"])
````

````python
#viewing our data top 5 after dropping the impact, month and day column

data.head()
````

````python
## dropping rows where time is NAN

data = data.dropna(subset = ["time"])
````

````python
## assigning a copy of our data to the data variable
data = data.copy()
````

````python
data.head()
````

````python
# importing the usd/jpy price data

data_price = pd.read_csv("./Data/DAT_ASCII_USDJPY_M1_2024.csv", sep=";", header=None, names=["Datetime", "Open", "High", "Low", "Close", "Volume"])
````

````python
#viewing top 5 of the usd/jpy price

data_price.head()
````

````python
#conveting our data to a datetime object

data_price["Datetime"] = pd.to_datetime(data_price["Datetime"], format="%Y%m%d %H%M%S")
````

````python
# sorting and making the datetime column the index from 0-end accordingly using the datetime column

data_price = data_price.set_index("Datetime").sort_index()
````

````python
 ## assigning the close datetime to match the data datetime and returns 0 if it does not match

def get_close_price(target_time):
    try:
        return data_price.loc[target_time, "Close"]
    except KeyError:
        return 0
````

````python
data_price.columns
````

````python
## dropping the high, low, open and volume column

data_price = data_price.drop(columns = ["High", "Low", "Open", "Volume"])
````

````python
data_price.head()
````

````python
data.head()
````

````python
#removing the day from the beginning in the date column

data["date"] = data["date"].str.replace(
    r"^(Mon|Tue|Wed|Thu|Fri|Sat|Sun)",
    "",
    regex=True
).str.strip()
````

````python
data.head()
````

````python
## creating a datetime column in the data

data["DateTime"] = pd.to_datetime(
    data["year"].astype(str) + " " +
    data["date"] + " " +
    data["time"],
    format="%Y %b %d %I:%M%p"
)
````

````python
## converting the news data from UTC +1 TO UTC

data["DateTime"] = (
    data["DateTime"]
    .dt.tz_localize("Etc/GMT-1")
    .dt.tz_convert("UTC")
)
````

````python
data.head()
````

````python
## removing the timezone information beside the datetime data

data["DateTime"] = data["DateTime"].dt.tz_localize(None)
````

````python
data.head()
````

````python
#if the next two days falls on a weekend, go to the next 2 or next 3 days

def get_close_48_72_96(news_time):

    for hrs in [48, 72, 96]:

        target = news_time + pd.Timedelta(hours=hrs)

        if target in data_price.index:
            return data_price.loc[target, "Close"]

    return 0
````

````python
#for after 24hrs that falls on weekend use 5 mins

def get_close_24h(news_time):
    # Try 24 hours after
    target_24 = news_time + pd.Timedelta(hours=24)

    if target_24 in data_price.index:
        return data_price.loc[target_24, "Close"]

    # Fallback: 5 minutes after
    target_5 = news_time + pd.Timedelta(minutes=5)

    if target_5 in data_price.index:
        return data_price.loc[target_5, "Close"]

    return 0
````

````python
## dropping the year, forecast, date and time column

data = data[[
    "DateTime",
    "currency",
    "event",
    "actual",
    "previous"
]]
````

````python
data.head()
````

````python
## creating the new columns for the price before, at and after news event

data.loc[:, "Close_5min_before"] = data["DateTime"].apply(
    lambda x: get_close_price(x - pd.Timedelta(minutes=5))
)

data.loc[:, "Close_at_release"] = data["DateTime"].apply(
    get_close_price
)

data.loc[:, "Close_5min_after"] = data["DateTime"].apply(
    lambda x: get_close_price(x + pd.Timedelta(minutes=5))
)

data.loc[:, "Close_24h_after"] = data["DateTime"].apply(
    get_close_24h
)


data.loc[:, "Close_48h_after"] = data["DateTime"].apply(
    get_close_48_72_96
)
````

````python
data.head(20)
````

````python
## doping actual or previos with NaN

data = data.dropna(subset = ["actual", "previous"])
````
